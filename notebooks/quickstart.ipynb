# Vexel quickstart on Colab / Kaggle

[Vexel](https://github.com/coolgamer203695-cyber/vexel) is a readable,
statically compiled language with a built-in tensor/ML engine — no
Python needed for the tensor parts. This notebook installs the
compiler on the VM and runs a demo.

> **Kaggle users:** turn **Settings → Internet → On** first,
> otherwise the clone/download cells cannot reach the network.

> **Note:** the VM is wiped when the session ends — re-run the
> install cell in each new session (it only takes a minute).

````python
# Install Node + Rust (only if missing) and the vexel CLI (~1-2 min).
!test -d vexel || git clone --depth 1 https://github.com/coolgamer203695-cyber/vexel
!bash vexel/tools/install_notebook.sh vexel
````

````python
!vexel version
````

````python
%%writefile quickstart.vxl
print "Hello from Vexel on this VM!"

import atlas

// tensors are built in - no Python required
w = tensor [0.5, -0.25, 1.0, 2.0]
print w.shape
print w.sum

// a fresh 4 -> 8 -> 2 MLP, forward pass on a rank-1 input
m = create model atlas.default_config 4, 8, 2
x = tensor [1, 2, 3, 4]
print (m.forward x).shape
print m.info.parameters

save model m to "demo.atlas"
m2 = load model "demo.atlas"
print m2.info.parameters
````

````python
!vexel run quickstart.vxl
````

Expected output:
```
Hello from Vexel on this VM!
[4]
3.25
[2]
58
58
```

What just happened: the compiler translated `quickstart.vxl` to
Rust, built it with `rustc`, and ran the native binary. The
`atlas` library gave you tensors, a fresh MLP model, a forward
pass, and an exact save/load round-trip of the native `.atlas`
format.

More: `vexel/run vexel/examples/` (hello, lists, http, model_demo,
nn_train), docs in `vexel/docs/atlas.md` — tensor ops, autograd,
and training loops (`loss.backward`, `optim.step`).
